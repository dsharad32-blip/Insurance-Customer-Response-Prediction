## **-** VISUALIZATION

### **1.** Target Variable Distribution

```python
plt.figure(figsize=(8,6))
ax = sns.countplot(
    data=df,
    x="Response",
    hue="Response",
    palette=["#0b5394", "#cc0000"]
)
ax.set_facecolor("#e6e6e6")

# Titles
plt.title("Customer Response Distribution", fontsize=18, weight="bold")
plt.xlabel("Response", fontsize=12)
plt.ylabel("Number of Customers", fontsize=12)

#Grid
ax.set_axisbelow(True)
plt.grid(True, linestyle="--", linewidth=1,alpha=0.7)

#bar labels
for container in ax.containers:
    ax.bar_label(container)

plt.show()
```

### **2.** Numerical Features vs Target 

```python
import matplotlib.pyplot as plt
import seaborn as sns

num_cols = ["Age", "Annual_Premium", "Vintage"]
for col in num_cols:

    plt.figure(figsize=(10,6))

    ax = sns.boxplot(
        data=df,
        x="Response",
        y=col,
        hue="Response",
        palette=["#0b5394", "#cc0000"]
    )

   
    ax.set_facecolor("#e6e6e6")

    # Grid
    ax.grid(True, color="white", linestyle="--", linewidth=1)

    # Titles
    ax.set_title(f"{col} vs Customer Response", fontsize=18, weight="bold")
    ax.set_xlabel("Response (0 = No, 1 = Yes)", fontsize=12)
    ax.set_ylabel(col, fontsize=12)

    # Remove extra borders
    sns.despine()

    plt.show()
```

### **3.** Violin Plot

```python
for col in num_cols:
    plt.figure(figsize=(10,6))
    ax = sns.violinplot(
        data=df,
        x="Response",
        y=col,
        hue="Response",
        palette=["#0b5394", "#cc0000"],
        legend=False
    )
  
ax.set_facecolor("#d9d9d9")
ax.grid(True, color="#bfbfbf", linestyle="--", linewidth=1,alpha=0.7)
ax.set_axisbelow(True)

ax.set_title(f"{col} Distribution by Customer Response",
                 fontsize=18, weight="bold")
ax.set_xlabel("Response", fontsize=12)
ax.set_ylabel(col, fontsize=12)
sns.despine()
plt.show()
```

### Categorical Variables vs Customer Response

```python
for col in cat_cols:

    plt.figure(figsize=(10,6))

    ax = sns.countplot(
        data=df,
        x=col,
        hue="Response",
        palette=["#0b5394", "#cc0000"]
    )

    
    ax.set_facecolor("#d9d9d9")
    ax.set_axisbelow(True)
    ax.yaxis.grid(True, color="white", linestyle="--", linewidth=1.2)
    ax.xaxis.grid(False)

    # Bar labels
    for container in ax.containers:
        ax.bar_label(container)

    # Titles
    ax.set_title(f"{col} vs Customer Response", fontsize=18, weight="bold")
    ax.set_xlabel(col, fontsize=12)
    ax.set_ylabel("Count", fontsize=12)

    plt.xticks(rotation=40)

    sns.despine()

    plt.show()
```

### Previously Insured vs Customer Response

```python
plt.figure(figsize=(10,6))

ax = sns.countplot(
    data=df,
    x="Previously_Insured",
    hue="Response",
    palette=["#0b5394", "#cc0000"]
)

# Light-dark grey background
ax.set_facecolor("#d9d9d9")

# Grid styling
ax.set_axisbelow(True)
ax.yaxis.grid(True, color="white", linestyle="--", linewidth=1.2)
ax.xaxis.grid(False)

# Bar labels
for container in ax.containers:
    ax.bar_label(container)

# Titles
ax.set_title("Previously Insured vs Customer Response", fontsize=18, weight="bold")
ax.set_xlabel("Previously Insured (0 = No, 1 = Yes)", fontsize=12)
ax.set_ylabel("Number of Customers", fontsize=12)

sns.despine()

plt.show()
```

##### CORRELATION-MATRIX

```python
# Correlation matrix
corr = df.corr(numeric_only=True)
cmap = sns.diverging_palette(130, 10, as_cmap=True)

plt.figure(figsize=(10,10))

ax = sns.heatmap(
    corr,
    annot=True,
    fmt=".2f",
    cmap=cmap,    
    square=True,
    linewidths=0.6,
    linecolor="white",
    cbar_kws={"shrink":0.8}
)

plt.title("Feature Correlation Matrix", fontsize=18, weight="bold", pad=20)

plt.xticks(rotation=90, ha="right", fontsize=11,weight="semibold")
plt.yticks(rotation=0, fontsize=11,weight="semibold")

plt.tight_layout()
plt.show()
```

### Annual Premium Distribution by Response

```python
plt.figure(figsize=(10,6))
ax = sns.boxplot(
    data=df,
    x="Response",
    y="Annual_Premium",
    hue="Response",
    palette=["#0b5394", "#cc0000"]
)

# Background
ax.set_facecolor("#e6e6e6")

# Grid
ax.set_axisbelow(True)
ax.yaxis.grid(True, linestyle="--", color="#bfbfbf")
ax.xaxis.grid(False)

plt.title("Annual Premium Distribution by Customer Response", fontsize=18, weight="bold")
plt.xlabel("Response (0 = No, 1 = Yes)")
plt.ylabel("Annual Premium")

plt.show()
```

### Age Groups vs Response Rate

```python
# Create Age Groups
df["Age_Group"] = pd.cut(
    df["Age"],
    bins=[18,25,35,50,65,100],
    labels=["18-25","26-35","36-50","51-65","65+"]
)
# Calculate response rate
age_response = pd.crosstab(df["Age_Group"], df["Response"], normalize="index")

# Plot
plt.figure(figsize=(24,14))

age_response.plot(
    kind="bar",
    stacked=True,
    colormap="crest"
)

plt.title("Customer Response Rate by Age Group", fontsize=18, weight="bold")
plt.xlabel("Age Group")
plt.ylabel("Proportion of Customers")

plt.grid(axis="y", linestyle="--", alpha=0.4)

plt.show()
```

### Previously Insured vs Customer Response (Detailed View)

```python
plt.figure(figsize=(8,5))
ax = sns.countplot(
    data=df,
    x="Previously_Insured",
    hue="Response",
    palette=["#0b5394", "#cc0000"]
)
# Background
ax.set_facecolor("#e6e6e6")

# Grid behind bars
ax.set_axisbelow(True)
ax.yaxis.grid(True, linestyle="--", color="#bfbfbf", alpha=0.8)
ax.xaxis.grid(False)

# Labels
plt.title("Previously Insured vs Customer Response", fontsize=18, weight="bold")
plt.xlabel("Previously Insured (0 = No, 1 = Yes)")
plt.ylabel("Number of Customers")

# Bar labels
for container in ax.containers:
    ax.bar_label(container)

plt.show()
```

### Key Factors Influencing Customer Response

```python
fig, axes = plt.subplots(2,2, figsize=(14,10))

fig.suptitle("Key Factors Influencing Customer Response", fontsize=20, weight="bold")

# Age vs Response (Mean Age)
age_mean = df.groupby("Response")["Age"].mean()

axes[0,0].bar(age_mean.index, age_mean.values, color=["#1f77b4","#d62728"])

axes[0,0].set_title("Average Age by Response")
axes[0,0].set_xlabel("Response")
axes[0,0].set_ylabel("Average Age")
axes[0,0].set_facecolor("#e6e6e6")
axes[0,0].grid(True, linestyle="--", alpha=0.5)


# Previously Insured vs Response (Stacked)
prev_resp = pd.crosstab(df["Previously_Insured"], df["Response"])

prev_resp.plot(kind="bar", stacked=True, ax=axes[0,1],
               color=["#1f77b4","#d62728"])

axes[0,1].set_title("Previously Insured vs Response")
axes[0,1].set_facecolor("#e6e6e6")
axes[0,1].grid(True, linestyle="--", alpha=0.5)


# Vehicle Age vs Response
veh_resp = pd.crosstab(df["Vehicle_Age"], df["Response"])

veh_resp.plot(kind="bar", ax=axes[1,0],
              color=["#1f77b4","#d62728"])

axes[1,0].set_title("Vehicle Age vs Response")
axes[1,0].set_facecolor("#e6e6e6")
axes[1,0].grid(True, linestyle="--", alpha=0.5)


# Vintage vs Response (Trend)
vintage_mean = df.groupby("Response")["Vintage"].mean()

axes[1,1].plot(vintage_mean.index,
               vintage_mean.values,
               marker="o",
               linewidth=3,
               color="#2ca02c")

axes[1,1].set_title("Customer Vintage Trend")
axes[1,1].set_xlabel("Response")
axes[1,1].set_ylabel("Average Vintage")
axes[1,1].set_facecolor("#e6e6e6")
axes[1,1].grid(True, linestyle="--", alpha=0.5)

plt.tight_layout(rect=[0,0,1,0.95])

plt.show()
```
