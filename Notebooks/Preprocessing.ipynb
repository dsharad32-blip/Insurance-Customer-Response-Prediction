### DATA PRE-PROCESSING
- Data Type Correction 
- Duplicate data handling
- Dropping Irrelevant Columns
- ***Objective*** : The goal of this step is to clean and prepare the raw Insurance_Customer_Response_Prediction dataset so it can be used effectively for **EDA** & **Model Training**.
  
- ****Output**** : A **Cleaned_dataset** ready for:EDA,Feature Engineering,ML Model Training.


```python
#Imports
import numpy as np
import pandas as pd
```

```python
#Load Data
DATA_PATH=r"C:\Users\dshar\OneDrive\Pictures\Desktop\Insurance_Customer_Response_Pred\DATA\data.csv"
RAW_DATA=pd.read_csv(DATA_PATH)
RAW_DATA.head()
```

```python
#Copy the data
df=RAW_DATA.copy()
df.head()
```

```python
# Info of Raw_data
df.info()
```

```python
#Meta Data
print("No of Rows:",df.shape[0])
print("No of Columns:",df.shape[1])
```

```python
# Column Categorization
all_columns=df.columns.to_list()
num_columns=df.select_dtypes(include="number").columns.to_list()
cat_columns=df.select_dtypes(include="object").columns.to_list()

#PRINTING
print("All Columns:\n",all_columns)
print("\nNumerical Columns:\n",num_columns)
print("\nCategorical Columns:\n",cat_columns)
```

```python
#Datatype of Columns
print("\nDtypes:\n",df.dtypes)
```

```python
#Checking Missing Values 
missing_values = df.isnull().sum()
print("\nMissing values:\n", missing_values)
```

```python
#Duplicate
duplicated_Val = df.duplicated().sum()
if duplicated_Val == 0:
    print("There are no duplicate values in the dataset.")
else:
    print(f"There are {duplicated_Val} duplicate values in the dataset.")
 
```

```python
#Dropping irrelevant Column(id)
df.drop(columns=["id"],axis=1,inplace=True)
```

```python
#Encoding the Column "Vehicle_Age"
df["Vehicle_Age"].unique()
print("Unique values in Vehicle_Age:", df["Vehicle_Age"].unique())

#Mapping dictionary for encoding
vehicle_age_map = {
    "< 1 Year": 0,
    "1-2 Year": 1,
    "> 2 Years": 2}
# Apply encoding
df["Vehicle_Age"] = df["Vehicle_Age"].map(vehicle_age_map).astype(int)
```

```python
#Checking Unique values 
print(df["Region_Code"].nunique())
print(df["Policy_Sales_Channel"].nunique())
```

```python
#Converting to Categorical Columns
df["Region_Code"] = df["Region_Code"].astype("object")
df["Policy_Sales_Channel"] = df["Policy_Sales_Channel"].astype("object")
```

```python
#Info after Cleaning 
df.info()
```

```python
#Saving Cleaned DAta 
CLEAN_DATA_PATH = r"C:\Users\dshar\OneDrive\Pictures\Desktop\Insurance_Customer_Response_Pred\DATA\Cleaned_data.csv"
df.to_csv(CLEAN_DATA_PATH, index=False)
print("Clean_data saved successfully.")
```
